Konsep dalam Probabilitas
1. Probabilitas
Definisi: 

Probabilitas suatu kejadian adalah rasio hasil yang menguntungkan terhadap jumlah total kemungkinan hasil.

Studi Kasus: Simulasi Pelemparan Dadu

Misalnya, probabilitas mendapatkan angka genap pada pelemparan dadu enam sisi adalah:

P(genap)=hasil menguntungkan / total hasil = 3/6 = 0.5 (50\%)

Implementasi Python:

```python
import numpy as np

np.random.seed(0)  # For reproducibility
throws = np.random.randint(1, 7, size=10000)  # Simulate 10,000 die rolls
even_count = np.sum(throws % 2 == 0)  # Count even numbers
probability_even = even_count / len(throws)
print(f"Simulated probability of rolling an even number: {probability_even}")
# Output: ~0.5085 (closer to 0.5 with more simulations)
```

Penjelasan: 

np.random.randint(1, 7, size=10000) menghasilkan 10.000 angka acak antara 1 dan 6 (simulasi pelemparan dadu).

pelemparan % 2 == 0 memeriksa apakah angka yang dihasilkan adalah genap (misalnya, 2, 4, atau 6).

np.sum() menghitung jumlah total hasil genap.

Probabilitas dihitung sebagai rasio jumlah hasil genap terhadap total percobaan. Semakin besar ukuran sampel, semakin dekat hasilnya ke nilai teoretis (0.5).

2. Probabilitas Bersyarat
Definisi: 

Probabilitas bersyarat adalah probabilitas kejadian B terjadi jika kejadian A sudah terjadi.

Studi Kasus: Probabilitas Angka Genap Jika Lebih dari 3

Jika hasil pelemparan dadu lebih dari 3, probabilitas bahwa angkanya genap adalah:

\( P (\text  {genap} \mid >3) = \frac{\text{hasil } (>3 \text{ dan genap})}{\text{total hasil } (>3)} = \frac{2}{3} \)

--------

3. Independensi Kejadian
Definisi: 

Dua kejadian dikatakan independen jika kejadian satu tidak memengaruhi probabilitas kejadian lainnya. Secara matematis:

\( P(A \cap B) = P(A) \cdot P(B) \)
Studi Kasus: Dua Dadu

Probabilitas mendapatkan angka genap pada dua dadu adalah:

\( P(\text{genap pada kedua dadu}) = P(\text{genap pada dadu 1}) \cdot P(\text{genap pada dadu 2}) = 0.5 \cdot 0.5 = 0.25 \text{ (25%)} \)
Selain itu, probabilitas salah satu kejadian A atau B terjadi adalah:

\( P(A \cup B) = P(A) + P(B) - P(A \cap B) \)

-----------

4. Kovariansi dan Independensi
Definisi: 

Kovariansi mengukur tingkat ketergantungan linier antara dua variabel acak. Jika dua variabel independen, kovariansinya akan nol. Namun, kovariansi nol tidak selalu menyiratkan independensi.

Implementasi Python:

```python
np.random.seed(0)
X = np.random.normal(loc=0, scale=1, size=1000)  # Dataset X
Y = np.random.normal(loc=0, scale=1, size=1000)  # Dataset Y

covariance_matrix = np.cov(X, Y)  # Covariance matrix
covariance = covariance_matrix[0, 1]  # Extract covariance
print(f"Covariance: {covariance}")
# Output: ~-0.0301 (close to zero, indicating near-independence)
```

Penjelasan: 

np.random.normal(loc=0, scale=1, size=1000) menghasilkan data dari distribusi normal dengan mean (loc) 0 dan deviasi standar (scale) 1.

np.cov(X, Y) menghitung matriks kovariansi antara dua dataset X dan Y.

Kovariansi diambil dari elemen [0, 1] matriks, yang menunjukkan hubungan linier antara X dan Y. Nilai mendekati nol berarti kedua variabel hampir independen.

Distribusi Probabilitas Umum
1. Distribusi Uniform
Definisi: 

Dalam distribusi uniform, semua hasil memiliki kemungkinan yang sama. Misalnya, probabilitas setiap hasil dalam lemparan koin adil adalah 0.5.

Implementasi Python:

```python
import matplotlib.pyplot as plt

np.random.seed(0)
uniform_data = np.random.uniform(0, 1, 1000)
plt.hist(uniform_data, bins=30, color='lightblue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()
```

Penjelasan: 

np.random.uniform(0, 1, 1000) menghasilkan 1.000 angka acak yang tersebar merata antara 0 dan 1.

plt.hist() membuat histogram untuk memvisualisasikan distribusi data. Histogram menunjukkan bahwa semua nilai memiliki frekuensi yang sama, sesuai dengan sifat distribusi uniform.

----------

2. Distribusi Binomial
Definisi: 

Distribusi binomial menggambarkan jumlah keberhasilan dalam sejumlah percobaan independen, masing-masing dengan probabilitas keberhasilan yang sama.

Implementasi Python:

```python
binomial_data = np.random.binomial(n=10, p=0.5, size=1000)
plt.hist(binomial_data, bins=30, color='lightblue', edgecolor='black')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()
```

Penjelasan: 

np.random.binomial(n=10, p=0.5, size=1000) mensimulasikan 1.000 eksperimen binomial dengan 10 percobaan dan probabilitas keberhasilan 0.5.

plt.hist() menampilkan distribusi hasil, yang cenderung berbentuk lonceng karena probabilitas keberhasilan sama (0.5).

------------

3. Distribusi Poisson
Definisi: 

Distribusi Poisson memodelkan jumlah kejadian yang terjadi dalam interval waktu atau ruang yang tetap.

Implementasi Python:

Hukum Bilangan Besar
Definisi: 

Hukum bilangan besar menyatakan bahwa semakin banyak observasi, rata-rata sampel akan mendekati nilai yang diharapkan.

Implementasi Python:

```python
sample_sizes = [10, 100, 1000, 10000]
means = []

for size in sample_sizes:
    sample = np.random.normal(loc=0, scale=1, size=size)
    means.append(np.mean(sample))

print(f"Rata-rata Sampel: {means}")
# Output: Rata-rata mendekati 0 seiring bertambahnya ukuran sampel
```
